```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from scikeras.wrappers import KerasRegressor
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import pandas as pd
```

```python
# ------------------------
# Step 0: Define Neural Net Model Builder
# ------------------------
def build_model(hidden_layer_1=64, hidden_layer_2=32, learning_rate=0.001):
    model = Sequential()
    model.add(Dense(hidden_layer_1, activation='relu', input_shape=(X1.shape[1],)))
    model.add(Dense(hidden_layer_2, activation='relu'))
    model.add(Dense(1))  # regression output
    model.compile(optimizer=Adam(learning_rate=learning_rate), loss='mean_squared_error')
    return model

# ------
```

```python
# Step 1: Create Pipeline
# ------------------------
pipeline = Pipeline(steps=[
    ('scaler', StandardScaler()),  # recommended for neural nets
    ('regressor', KerasRegressor(model=build_model, verbose=0, random_state=42))
])
```

```python
param_grid = {
    'regressor__model__hidden_layer_1': [64,128],
    'regressor__model__hidden_layer_2': [32,64],
    'regressor__model__learning_rate': [0.001, 0.01],
    'regressor__batch_size': [32,64],
    'regressor__epochs': [3]
}
```

```python
# ------------------------
# Step 3: GridSearchCV
# ------------------------
search = RandomizedSearchCV(
    pipeline,
    param_distributions=param_grid,
    n_iter=10,
    cv=3,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1,
    verbose=2
)
# --
```
